```python
import os
import pandas as pd

path1 = 'dataset/PIHPS_Jakarta_Full_2021_2026.csv'
path2 = 'PIHPS_Jakarta_Full_2021_2026.csv'

if os.path.exists(path1):
    path = path1
elif os.path.exists(path2):
    path = path2
else:
    raise FileNotFoundError("Neither file path exists.")

df_harga = pd.read_csv(path, sep=',')
df_harga.info()
```

```python
# ubah format tanggal
df_harga["Tanggal"] = pd.to_datetime(df_harga["Tanggal"], format="%Y-%m-%d")

df_harga.info()
```

## Bersihkan format harga


```python
# Ubah kolom harga dari string ke numeric
harga_columns = df_harga.columns[1:]  # Asumsikan kolom pertama adalah 'Tanggal'
for col in harga_columns:
    df_harga[col] = df_harga[col].str.replace(',', '', regex=False)  # Hapus koma sebagai pemisah ribuan
    df_harga[col] = pd.to_numeric(df_harga[col], errors='coerce')  # Ubah ke numeric, non-numeric akan menjadi NaN
df_harga.info()
```

## Visualisasi distribusi harga

```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

sns.set(style="whitegrid", palette="muted", font_scale=1.1)

# Siapkan data numerik dan time series
harga_numeric = df_harga.select_dtypes(include=[np.number])
harga_time = df_harga.set_index("Tanggal")[harga_numeric.columns]

# Visualisasi tren harga beberapa komoditas utama
selected = [
    "Beras",
    "Minyak Goreng",
    "Gula Pasir",
    "Telur Ayam",
    "Cabai Merah",
    "Cabai Rawit",
    "Daging Ayam"
]
selected = [c for c in selected if c in harga_time.columns]

plt.figure(figsize=(14, 6))
for col in selected:
    plt.plot(harga_time.index, harga_time[col], label=col)

plt.legend(loc="upper left")
plt.title("Tren harga komoditas utama")
plt.xlabel("Tanggal")
plt.ylabel("Harga (Rp)")
plt.tight_layout()
plt.show()

# Plot distribusi harga untuk contoh komoditas
example_cols = [c for c in ["Beras", "Cabai Merah", "Minyak Goreng", "Telur Ayam"] if c in harga_numeric.columns]
fig, axes = plt.subplots(len(example_cols), 2, figsize=(14, 4 * len(example_cols)))
for i, col in enumerate(example_cols):
    sns.histplot(harga_numeric[col].dropna(), kde=True, ax=axes[i, 0], color="tab:blue")
    axes[i, 0].set_title(f"Histogram {col}")
    sns.boxplot(x=harga_numeric[col], ax=axes[i, 1], color="tab:orange")
    axes[i, 1].set_title(f"Boxplot {col}")
plt.tight_layout()
plt.show()

# Korelasi antar komoditas
corr = harga_numeric.corr()
plt.figure(figsize=(16, 14))
sns.heatmap(corr, cmap="coolwarm", center=0, linewidths=.5, annot=False)
plt.title("Heatmap korelasi antar komoditas")
plt.show()

# Analisis outlier sederhana menggunakan IQR
q1 = harga_numeric.quantile(0.25)
q3 = harga_numeric.quantile(0.75)
iqr = q3 - q1
lower = q1 - 1.5 * iqr
upper = q3 + 1.5 * iqr
outlier_counts = ((harga_numeric < lower) | (harga_numeric > upper)).sum().sort_values(ascending=False)
print("Jumlah outlier per komoditas menurut IQR:\n", outlier_counts.head(15))

# Rata-rata bulanan untuk komoditas terpilih
monthly = harga_time.resample("ME").mean()
selected_monthly = [c for c in ["Beras", "Minyak Goreng", "Gula Pasir", "Telur Ayam"] if c in monthly.columns]

plt.figure(figsize=(14, 6))
for col in selected_monthly:
    plt.plot(monthly.index, monthly[col], marker="o", label=col)

plt.legend(loc="upper left")
plt.title("Rata-rata harga bulanan komoditas terpilih")
plt.xlabel("Bulan")
plt.ylabel("Harga rata-rata (Rp)")
plt.tight_layout()
plt.show()

# Tampilkan ringkasan tambahan untuk verifikasi
print("Jumlah baris:", len(df_harga))
print("Rentang tanggal:", df_harga["Tanggal"].min(), "sampai", df_harga["Tanggal"].max())
print("Jumlah missing per komoditas (top 10):\n", df_harga.isna().sum().sort_values(ascending=False).head(10))
```
